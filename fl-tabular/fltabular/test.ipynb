```python
import os
import sys

sys.path.insert(0, "..")

import torch
import pandas as pd
from fltabular.task import CostRegressor, evaluator, get_input_dim, load_data

os.environ["FL_TABULAR_TRAIN_DATA_PATH"] = "../train_db.pkl"
os.environ["FL_TABULAR_VAL_DATA_PATH"] = "../val_db.pkl"
os.environ["FL_TABULAR_TARGET_COLUMN"] = "COST_BL"

model = CostRegressor(get_input_dim())
state_dict = torch.load("../final_model.pt", map_location="cpu")
model.load_state_dict(state_dict)

test_loader = load_data("val", partition_id=0, num_partitions=1)
result = evaluator(model, test_loader)
mae, rmse, mse, r2 = result[0], result[2], result[1], result[3]
print("MAE:", mae)
print("RMSE:", rmse)
print("MSE:", mse)
print("R²:", r2)
```
